```python
import csv
from IPython.display import display, HTML

# 2. Load students_marks.csv and calculate average for each student
file_path = '/content/students_marks.csv'
students_list = []

with open(file_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        # Identify the 5 subject scores dynamically or by filtering out non-subject columns
        scores = []
        for key, val in row.items():
            if key.lower() not in ['name', 'rollno', 'roll_no', 'roll number', 'section', 'id', 'student_id']:
                try:
                    scores.append(float(val))
                except ValueError:
                    pass

        # Add Average key (mean of the five subjects, 1 decimal)
        if scores:
            row['Average'] = round(sum(scores) / len(scores), 1)
        else:
            row['Average'] = 0.0

        students_list.append(row)

# Sort students by Average in descending order to find the top 5
top_5_students = sorted(students_list, key=lambda x: x['Average'], reverse=True)[:5]


```

```python
# 3. Build results.html content
html_content = f"""<!DOCTYPE html>
<html>
<head>
    <title>Student Performance Dashboard</title>
    <style>
        body {{ font-family: Arial, sans-serif; margin: 20px; }}
        table {{ border-collapse: collapse; width: 60%; margin-top: 20px; }}
        th, td {{ border: 1px solid #ddd; padding: 8px; text-align: left; }}
        th {{ background-color: #f2f2f2; }}
        .highlight {{ font-weight: bold; color: #2c3e50; }}
    </style>
</head>
<body>
    <h1>Student Performance Report</h1>
    <p>Data loaded from the source file: <span class="highlight">{file_path}</span></p>

    <h2>Dashboard Contents</h2>
    <ul>
        <li>Overall performance average per student</li>
        <li>Ranked leaderboard of academic achievement</li>
        <li>Detailed overview of the top 5 performing students</li>
    </ul>

    <h2>Top 5 Students Leaderboard</h2>
    <table>
        <thead>
            <tr>
                <th>RollNo</th>
                <th>Name</th>
                <th>Section</th>
                <th>Average</th>
            </tr>
        </thead>
        <tbody>
"""

for student in top_5_students:
    roll_no = student.get('RollNo') or student.get('rollno') or student.get('Roll No') or 'N/A'
    name = student.get('Name') or 'N/A'
    section = student.get('Section') or 'N/A'
    average = student.get('Average', 0.0)
    html_content += f"""            <tr>
                <td>{roll_no}</td>
                <td>{name}</td>
                <td>{section}</td>
                <td>{average}</td>
            </tr>
"""

html_content += """        </tbody>
    </table>
</body>
</html>
"""
```

```python
# 4. Save the file as results.html
output_html_path = 'results.html'
with open(output_html_path, 'w', encoding='utf-8') as f:
    f.write(html_content)

# Show the HTML file content inside the notebook
display(HTML(html_content))
```

```python
# Print the names and averages of the top 5 students
print("\nTop 5 Students by Average:")
print("---------------------------")
for idx, student in enumerate(top_5_students, 1):
    name = student.get('Name') or 'N/A'
    average = student.get('Average', 0.0)
    print(f"{idx}. {name} - Average: {average}")
```

**Lab Task 2**:**CSS classes and ids**

```python
import csv
from IPython.display import display, HTML

# 1. Load and calculate average for all students
file_path = '/content/students_marks.csv'
students_list = []

with open(file_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        scores = []
        for key, val in row.items():
            if key.lower() not in ['name', 'rollno', 'roll_no', 'roll number', 'section', 'id', 'student_id']:
                try:
                    scores.append(float(val))
                except ValueError:
                    pass

        if scores:
            row['Average'] = round(sum(scores) / len(scores), 1)
        else:
            row['Average'] = 0.0
        students_list.append(row)

# Sort all 25 students by Average descending
sorted_students = sorted(students_list, key=lambda x: x['Average'], reverse=True)



```

```python
# 2 & 3. Build HTML and apply CSS styling
html_style = """
<style>
    table {
        border-collapse: collapse;
        width: 70%;
        font-family: Arial, sans-serif;
        margin-top: 20px;
    }
    th, td {
        border: 1px solid #ddd;
        padding: 8px;
        text-align: left;
    }
    tr.header-row {
        background-color: #2c3e50;
        color: white;
    }
    .low {
        color: red;
        font-weight: bold;
    }
    #topper {
        background-color: gold !important;
    }
</style>
"""

html_table = html_style + """
<h2>All Students Ranked by Average</h2>
<table>
    <thead>
        <tr class="header-row">
            <th>#</th>
            <th>Name</th>
            <th>Section</th>
            <th>Average</th>
        </tr>
    </thead>
    <tbody>
"""

for idx, student in enumerate(sorted_students, 1):
    name = student.get('Name') or 'N/A'
    section = student.get('Section') or 'N/A'
    average = student.get('Average', 0.0)

    # Determine row class and id attributes
    row_id_attr = ' id="topper"' if idx == 1 else ''
    row_class_attr = ' class="low"' if average < 60.0 else ' class="ok"'

    html_table += f"""        <tr{row_id_attr}{row_class_attr}>
            <td>{idx}</td>
            <td>{name}</td>
            <td>{section}</td>
            <td>{average}</td>
        </tr>
"""

html_table += """    </tbody>
</table>
"""

# Save the rendered HTML to file and display
with open('students_leaderboard.html', 'w', encoding='utf-8') as f:
    f.write(html_table)

display(HTML(html_table))
```

```python
# 4. Print stats
below_60_students = [s for s in sorted_students if s['Average'] < 60.0]
topper = sorted_students[0]

print("\n================ STATISTICS ================")
print(f"Students below 60 ({len(below_60_students)} total):")
for s in below_60_students:
    print(f"  - {s['Name']} (Average: {s['Average']})")

print(f"\nTopper of the class:")
print(f"  Name: {topper['Name']}")
print(f"  Average: {topper['Average']}")
```

**Lab Task 3:** **The placeholder dashboard**

```python
import os
from IPython.display import HTML, display

# 1. Create the dashboard folder
os.makedirs("dashboard", exist_ok=True)

# Compute the stats dynamically in Python
total_students = len(students_list)
class_avg = round(sum(s['Average'] for s in students_list) / total_students, 1) if total_students > 0 else 0.0
passed_count = sum(1 for s in students_list if s['Average'] >= 50.0)
topper_name = sorted_students[0]['Name'] if sorted_students else "N/A"

# Define the HTML template
html = """<!DOCTYPE html>
<meta charset="utf-8">
<link rel="stylesheet" href="style.css">
<body>
  <div id="chart-holder" class="dev">
    <div id="header">
      <h2>BSDS Results Dashboard</h2>
      <p>This interactive dashboard displays overall academic performance, subject distributions, and class leaderboards for the BSCS (Data Science) students.</p>     </div>
    <div id="chart-components">
      <div id="main">A placeholder for the main chart.</div>
      <div id="sidebar">
        <h3>Summary Stats</h3>
        <p class="stat">Total Students: <strong id="stat-total">""" + str(total_students) + """</strong></p>
        <p class="stat">Class Average: <strong id="stat-avg">""" + str(class_avg) + """</strong></p>
        <p class="stat">Passed (>=50): <strong id="stat-passed">""" + str(passed_count) + """</strong></p>
        <p class="stat">Topper: <strong id="stat-topper">""" + topper_name + """</strong></p>
      </div>
    </div>
  </div>
  <script src="script.js"></script>
</body>"""

# Define CSS for styling and layout (720px width and 320px height for chart-components)
css = """
body { background: #ccc; font-family: sans-serif; }

div.dev { border: solid 1px red; }
div.dev div { border: dashed 1px green; }

div#chart-holder { width: 720px; background: white; margin: auto; }
div#chart-components { height: 320px; position: relative; }
div#main, div#sidebar { position: absolute; }

div#main { width: 75%; height: 100%; background: #eee; }
div#sidebar { right: 0; width: 25%; height: 100%; padding: 10px; box-sizing: border-box; }
.stat { margin: 8px 0; font-size: 14px; }
"""

# Define JavaScript
js = """console.log("Dashboard metrics loaded successfully.");"""

# Write files into the dashboard directory
for filename, content in [("index.html", html), ("style.css", css), ("script.js", js)]:
    with open(os.path.join("dashboard", filename), "w", encoding="utf-8") as f:
        f.write(content)

# Define function to inject css and js styles into index.html and show the dashboard
def show(folder):
    with open(os.path.join(folder, "index.html"), "r", encoding="utf-8") as f:
        page = f.read()
    with open(os.path.join(folder, "style.css"), "r", encoding="utf-8") as f:
        style_content = f.read()
    with open(os.path.join(folder, "script.js"), "r", encoding="utf-8") as f:
        script_content = f.read()

    page = page.replace('<link rel="stylesheet" href="style.css">', "<style>" + style_content + "</style>")
    page = page.replace('<script src="script.js"></script>', "<script>" + script_content + "</script>")
    display(HTML(page))

# Render dashboard and list its files
show("dashboard")

print("Summary Statistics:")
print(f"Total Students: {total_students}")
print(f"Class Average: {class_avg}")
print(f"Passed (>=50): {passed_count}")
print(f"Topper: {topper_name}")
```

**Lab Task 4:** **A hand-made SVG line chart**

```python
subjects = ['Math', 'Stats', 'Python', 'Databases', 'English']
subject_averages = {}

for sub in subjects:
    # Calculate the average score for each subject from students_list
    scores = [float(student[sub]) for student in students_list if sub in student]
    subject_averages[sub] = round(sum(scores) / len(scores), 1) if scores else 0.0

print("Class Averages for Each Subject:")
print("-------------------------------")
for sub, avg in subject_averages.items():
    print(f"{sub}: {avg}")
```

```python
# Calculate coordinates using the new formula: y = 180 - (mark - 40) * 3.5
# Note: With this formula, a mark of 40 corresponds to y = 180.
x_coords = [55 + i * 62.5 for i in range(5)]
y_coords = [180 - (subject_averages[sub] - 40) * 3.5 for sub in subjects]

# Construct the path data (d attribute)
path_d = f"M {x_coords[0]:.1f} {y_coords[0]:.1f} " + " ".join(f"L {x:.1f} {y:.1f}" for x, y in zip(x_coords[1:], y_coords[1:]))

print("Path's d text:")
print(path_d)
print("\n")

# Define the SVG with styles separated into a <style> block
svg_content = f"""<svg width="380" height="210" class="chart-container">
  <style>
    .chart-container {{ background: white; border: 1px solid #ccc; font-family: sans-serif; }}
    .title {{ font-size: 14px; font-weight: bold; text-anchor: middle; fill: #2c3e50; }}
    .axis-line {{ stroke: #333; stroke-width: 1.5; }}
    .grid-line {{ stroke: #ddd; stroke-dasharray: 3,3; }}
    .axis-label {{ font-size: 10px; text-anchor: end; fill: #555; }}
    .rotated-label {{ font-size: 11px; font-weight: bold; text-anchor: middle; fill: #555; }}
    .legend-box {{ fill: #fafafa; stroke: #ccc; stroke-width: 1; rx: 3px; }}
    .legend-line {{ stroke: #3498db; stroke-width: 2.5; }}
    .legend-text {{ font-size: 9px; fill: #333; }}
    .chart-path {{ fill: none; stroke: #3498db; stroke-width: 2.5; }}
    .data-point {{ fill: #e74c3c; stroke: #fff; stroke-width: 1.5; }}
    .value-label {{ font-size: 9px; text-anchor: middle; font-weight: bold; fill: #2c3e50; }}
    .x-label {{ font-size: 10px; text-anchor: middle; fill: #555; }}
  </style>

  <!-- Centered Title -->
  <text x="190" y="22" class="title">Class Averages by Subject</text>

  <!-- Axes -->
  <line x1="55" y1="180" x2="305" y2="180" class="axis-line" />
  <line x1="55" y1="40" x2="55" y2="180" class="axis-line" />

  <!-- Y-Axis Grid Lines & Ticks (40, 60, 80, 100) based on new scale -->
  <line x1="50" y1="180" x2="305" y2="180" class="grid-line" />
  <text x="45" y="183" class="axis-label">40</text>

  <line x1="50" y1="110" x2="305" y2="110" class="grid-line" />
  <text x="45" y="114" class="axis-label">60</text>

  <line x1="50" y1="40" x2="305" y2="40" class="grid-line" />
  <text x="45" y="44" class="axis-label">80</text>

  <!-- Y-Axis Label Rotated -90 -->
  <text x="-110" y="16" transform="rotate(-90)" class="rotated-label">Average Score</text>

  <!-- Legend Box -->
  <rect x="310" y="40" width="62" height="35" class="legend-box" />
  <line x1="315" y1="57" x2="330" y2="57" class="legend-line" />
  <text x="335" y="60" class="legend-text">Class</text>

  <!-- Connecting Path Line -->
  <path d="{path_d}" class="chart-path" />
"""

# Add Circles & Values dynamically
for x, y, sub in zip(x_coords, y_coords, subjects):
    avg = subject_averages[sub]
    svg_content += f"""  <!-- Data Point -->
  <circle cx="{x:.1f}" cy="{y:.1f}" r="4.5" class="data-point" />
  <text x="{x:.1f}" y="{y-8:.1f}" class="value-label">{avg}</text>
  <!-- X-Axis Labels -->
  <text x="{x:.1f}" y="196" class="x-label">{sub}</text>
"""

svg_content += "</svg>"

# Display the SVG inside the notebook
display(HTML(svg_content))
```

**Lab Task 5:** Fill the main box with a bar chart

```python
import os
from IPython.display import HTML, display

# 1. Setup coordinates and variables for the SVG Bar Chart
# SVG Dimensions: 520 x 300
# Margins: Left=40, Right=20, Top=30, Bottom=70
# Plot width = 520 - 40 - 20 = 460
# Plot height = 300 - 30 - 70 = 200
# X scale: 25 bars -> width per bar = 18px (14px bar + 4px gap)
# Y scale: 0 to 100 mapping to 0 to 200px height. Scale factor = 2.0
# y_coord = 200 - (score * 2.0)

bar_width = 14
gap = 4
scale_y = 2.0
plot_height = 200

svg_bars = """<svg width="520" height="300" class="bar-chart">
  <g transform="translate(40, 30)">
    <!-- Grid lines -->
    <line x1="0" y1="200" x2="450" y2="200" class="axis-line" />
    <line x1="0" y1="0" x2="0" y2="200" class="axis-line" />

    <line x1="0" y1="120" x2="450" y2="120" class="grid-line" />
    <text x="-8" y="123" class="axis-label">40</text>

    <line x1="0" y1="80" x2="450" y2="80" class="grid-line" />
    <text x="-8" y="83" class="axis-label">60</text>

    <line x1="0" y1="40" x2="450" y2="40" class="grid-line" />
    <text x="-8" y="43" class="axis-label">80</text>

    <line x1="0" y1="0" x2="450" y2="0" class="grid-line" />
    <text x="-8" y="3" class="axis-label">100</text>

    <!-- Threshold dashed line at 60 (y = 200 - 60*2 = 80) -->
    <line x1="0" y1="80" x2="450" y2="80" class="threshold-line" />
"""

# Populate SVG bars and labels dynamically
for idx, student in enumerate(sorted_students):
    name = student.get('Name', 'N/A')
    first_name = name.split()[0] if name else 'N/A'
    avg = student.get('Average', 0.0)

    # Coordinates
    x = idx * (bar_width + gap)
    h = avg * scale_y
    y = plot_height - h

    bar_class = "bar low" if avg < 60.0 else "bar normal"

    svg_bars += f"""    <!-- Bar for {name} -->
    <rect x="{x}" y="{y}" width="{bar_width}" height="{h}" class="{bar_class}" />
    <!-- Rotated label for first name -->
    <text x="{x + bar_width/2}" y="210" class="student-label" transform="rotate(-60, {x + bar_width/2}, 210)">{first_name}</text>
"""

svg_bars += """  </g>
</svg>"""

# 2. Update dashboard/index.html with the dynamic bar chart inside <div id="main">
html = f"""<!DOCTYPE html>
<meta charset="utf-8">
<link rel="stylesheet" href="style.css">
<body>
  <div id="chart-holder" class="dev">
    <div id="header">
      <h2>BSDS Results Dashboard</h2>
      <p>This interactive dashboard displays overall academic performance, subject distributions, and class leaderboards for the BSCS (Data Science) students.</p>
    </div>
    <div id="chart-components">
      <div id="main">{svg_bars}</div>
      <div id="sidebar">
        <h3>Summary Stats</h3>
        <p class="stat">Total Students: <strong id="stat-total">{total_students}</strong></p>
        <p class="stat">Class Average: <strong id="stat-avg">{class_avg}</strong></p>
        <p class="stat">Passed (>=50): <strong id="stat-passed">{passed_count}</strong></p>
        <p class="stat">Topper: <strong id="stat-topper">{topper_name}</strong></p>
      </div>
    </div>
  </div>
  <script src="script.js"></script>
</body>"""

# 3. Update style.css with the chart styles
css = """
body { background: #ccc; font-family: sans-serif; }

div.dev { border: solid 1px red; }
div.dev div { border: dashed 1px green; }

div#chart-holder { width: 720px; background: white; margin: auto; }
div#chart-components { height: 320px; position: relative; }
div#main, div#sidebar { position: absolute; }

div#main { width: 75%; height: 100%; background: #eee; }
div#sidebar { right: 0; width: 25%; height: 100%; padding: 10px; box-sizing: border-box; }
.stat { margin: 8px 0; font-size: 14px; }

/* SVG Bar Chart Styling */
.bar-chart { width: 100%; height: 100%; }
.axis-line { stroke: #333; stroke-width: 1.5px; }
.grid-line { stroke: #ddd; stroke-width: 1px; stroke-dasharray: 2,2; }
.axis-label { font-size: 9px; fill: #444; text-anchor: end; }
.threshold-line { stroke: red; stroke-width: 1.5px; stroke-dasharray: 4,4; }
.bar { stroke: #fff; stroke-width: 0.5px; }
.bar.normal { fill: #008080; } /* Teal */
.bar.low { fill: red; } /* Red for below 60 */
.student-label { font-size: 8px; fill: #333; text-anchor: end; }
""".strip()

# 4. Save and re-render
with open(os.path.join("dashboard", "index.html"), "w", encoding="utf-8") as f:
    f.write(html)
with open(os.path.join("dashboard", "style.css"), "w", encoding="utf-8") as f:
    f.write(css)

show("dashboard")
```
