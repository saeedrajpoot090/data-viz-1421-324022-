**TASK 5:**

1. **The number of students in each of six classes.**

    **Chart:** Bar Chart
    **Why:** A bar chart is ideal for comparing discrete, categorical groups (the six classes) against a numerical value (number of students).

2. **The price of petrol every month for two years.**
   
    **Chart:** Line Chart
    **Why:** A line chart is the standard choice for visualizing trends, changes, and fluctuations in a continuous variable over a continuous time interval.

3. **The height and weight of 60 students, to see whether taller students weigh more.**
   
    **Chart:** Scatter Plot
   **Why:** A scatter plot is perfect for displaying individual data points to identify relationships, correlations, or patterns between two numerical variables (height and weight).

4. **How a Rs. 50,000 monthly income is divided between rent, food, transport and savings.**
   
    **Chart:** Pie Chart
    **Why:** A pie chart effectively displays part-to-whole relationships, showing how a single total amount is distributed among a small number of categories.

5. **How the exam marks of 200 students are spread between 0 and 100.**
   
    **Chart:** Histogram
    **Why:** A histogram groups continuous numerical data into bins to show the frequency distribution and spread of scores across different ranges.

6. **Sales of three shops in four different quarters, to compare the shops.**
   
    **Chart:** Grouped Bar Chart
    **Why:** A grouped bar chart allows you to easily compare discrete subcategories (the three shops) grouped within broader categories (the four quarters).
